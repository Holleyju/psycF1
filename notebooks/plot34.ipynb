```python
import sys
from pathlib import Path
import pandas as pd

sys.path.append("../src")
from f1_pipeline.plotting import prepare_team_stops, plot_team_boxplot, plot_pace_control

raw = Path("../data/raw")
processed = Path("../data/processed")
figures = Path("../figures")
```

```python
races = pd.read_csv(raw / "races.csv", na_values="\\N")
constructors = pd.read_csv(raw / "constructors.csv", na_values="\\N")
good_ids = pd.read_csv(processed / "good_races.csv")["raceId"].tolist()
results_clean = pd.read_csv(processed / "results_clean.csv")
pits_clean = pd.read_csv(processed / "pit_stops_clean.csv")
table = pd.read_csv(processed / "driver_race_table.csv")
```

```python
stops_2023 = prepare_team_stops(pits_clean, results_clean, constructors, good_ids, 2023, races)
plot_team_boxplot(stops_2023, "2023: pit stop speed and consistency by team",
                  save_path=figures / "plot3_team_box_2023");
```

```python
plot_pace_control(table, "Does pit speed still matter at equal car pace?",
                  save_path=figures / "plot4_pace_control");
```
