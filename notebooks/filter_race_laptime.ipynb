```python
import sys
from pathlib import Path
import pandas as pd

sys.path.append("../src") 
from f1_pipeline.filter_races import get_good_race_ids
```

```python
raw = Path("../data/raw")
processed = Path("../data/processed")

races = pd.read_csv(raw / "races.csv", na_values="\\N")
laps = pd.read_csv(raw / "lap_times.csv", na_values="\\N")
```

```python
good_ids = get_good_race_ids(races, laps)
```

```python
good_races = races[races["raceId"].isin(good_ids)]
good_races.to_csv(processed / "good_races.csv", index=False)
```
