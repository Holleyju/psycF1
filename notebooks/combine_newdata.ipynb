```python
import sys
from pathlib import Path
import pandas as pd

sys.path.append("../src")  
from combine import build_driver_race_table
```

```python
raw = Path("../data/raw")
processed = Path("../data/processed")

good_races = pd.read_csv(processed / "good_races.csv")
results_clean = pd.read_csv(processed / "results_clean.csv")
pits_clean = pd.read_csv(processed / "pit_stops_clean.csv")
laps = pd.read_csv(raw / "lap_times.csv", na_values="\\N")
```

```python
table = build_driver_race_table(good_races, results_clean, pits_clean, laps)
```

```python
table = table.rename(columns={
    "name": "race_name",
    "constructorId": "team_id",
    "grid": "start_position",
    "positionOrder": "final_rank",
})
table["start_position"] = table["start_position"].astype(int)
table["positions_gained"] = table["start_position"] - table["final_rank"]
```

```python
print(table.shape)
table.head(10)
```

```python
table.to_csv(processed / "driver_race_table.csv", index=False)
print("Saved to", processed / "driver_race_table.csv")
```
